```python

from azure.ai.ml import MLClient
from azure.identity import DefaultAzureCredential

subscription_id = "d612bb45-6c1e-4b93-ac48-9b0d75744120"
resource_group = "rg-diabetes-mlops"
workspace_name = "mlflow-workspace"

credential = DefaultAzureCredential()

ml_client = MLClient(
    credential=credential,
    subscription_id=subscription_id,
    resource_group_name=resource_group,
    workspace_name=workspace_name,
)

workspace = ml_client.workspaces.get(workspace_name)

print("Connected to Azure ML successfully!")
print("Workspace:", workspace.name)
print("Resource group:", resource_group)
```

```python

from pathlib import Path
from azure.ai.ml.entities import Model
from azure.ai.ml.constants import AssetTypes

# Project directory
PROJECT_DIR = Path.home() / "cloudfiles" / "code" / "capstone-project-1-fraud"

# Verify the required files exist
required_files = ["model.joblib", "features.json", "metadata.json"]

for filename in required_files:
    file_path = PROJECT_DIR / filename
    print(filename, "FOUND" if file_path.exists() else "MISSING")

# Register only if all required files exist
if all((PROJECT_DIR / f).exists() for f in required_files):
    registered_model = ml_client.models.create_or_update(
        Model(
            path=str(PROJECT_DIR),
            name="online-fraud-smote-rf",
            type=AssetTypes.CUSTOM_MODEL,
            description=(
                "Online fraud detection Random Forest pipeline "
                "trained with SMOTE. Includes feature and metadata files."
            ),
        )
    )

    print("\nModel registration successful!")
    print("Name:", registered_model.name)
    print("Version:", registered_model.version)
    print("Asset ID:", registered_model.id)
else:
    print("\nRegistration skipped. Please check the missing files.")
```
